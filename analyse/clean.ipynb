# Nettoyage des données - Suppression des colonnes avec valeurs manquantes

Ce notebook charge les données de `friches-standard.csv`, supprime les colonnes avec trop de valeurs manquantes, et exporte le résultat nettoyé.

## 1. Importation des bibliothèques

```python
# Importation des bibliothèques nécessaires
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Bibliothèques importées avec succès !")
```

## 2. Chargement des données

```python
# Chargement des données depuis le fichier CSV
df = pd.read_csv('../data/friches-standard.csv', sep=';', encoding='utf-8', low_memory=False)

print(f"Données chargées avec succès !")
print(f"Nombre de lignes : {df.shape[0]:,}")
print(f"Nombre de colonnes : {df.shape[1]}")
```

## 3. Suppression des colonnes avec trop de valeurs manquantes

```python
# Liste des colonnes à supprimer (celles avec beaucoup de données manquantes)
colonnes_a_supprimer = [
    'geomsurf',
    'desserte_commentaire',
    'site_ademe_url',
    'desserte_distance',
    'site_projet_url',
    'site_reconv_annee',
    'sol_pollution_commentaire',
    'sol_pollution_annee',
    'sol_depollution_fiche',
    'bati_surface',
    'activite_code',
    'source_contact',
    'site_reconv_type',
    'activite_fin_annee',
    'urba_zone_formdomi',
    'site_adresse'
]

print(f"📊 Dataset AVANT nettoyage :")
print(f"   - Nombre de lignes : {df.shape[0]:,}")
print(f"   - Nombre de colonnes : {df.shape[1]}")
print(f"\n🗑️  Colonnes à supprimer : {len(colonnes_a_supprimer)}")

# Suppression des colonnes
df_clean = df.drop(columns=colonnes_a_supprimer, errors='ignore')

print(f"\n✅ Dataset APRÈS nettoyage :")
print(f"   - Nombre de lignes : {df_clean.shape[0]:,}")
print(f"   - Nombre de colonnes : {df_clean.shape[1]}")
print(f"   - Colonnes supprimées : {df.shape[1] - df_clean.shape[1]}")
```

## 4. Vérification des colonnes restantes

```python
# Liste des colonnes restantes
print("=== Colonnes restantes dans le dataset nettoyé ===\n")
for i, col in enumerate(df_clean.columns, 1):
    print(f"{i:2d}. {col}")
```

```python
# Analyse des valeurs manquantes restantes
missing_data_clean = pd.DataFrame({
    'Colonne': df_clean.columns,
    'Valeurs_manquantes': df_clean.isnull().sum(),
    'Pourcentage': (df_clean.isnull().sum() / len(df_clean) * 100).round(2)
})
missing_data_clean = missing_data_clean[missing_data_clean['Valeurs_manquantes'] > 0].sort_values('Pourcentage', ascending=False)

print(f"\n=== Valeurs manquantes restantes (sur {len(df_clean.columns)} colonnes) ===\n")
if len(missing_data_clean) > 0:
    print(missing_data_clean.to_string(index=False))
    print(f"\n⚠️  Il reste {len(missing_data_clean)} colonnes avec des valeurs manquantes")
else:
    print("✅ Aucune valeur manquante !")
```

```python
# Affichage de TOUTES les colonnes (avec et sans valeurs manquantes)
all_columns_stats = pd.DataFrame({
    'Colonne': df_clean.columns,
    'Valeurs_manquantes': df_clean.isnull().sum(),
    'Pourcentage': (df_clean.isnull().sum() / len(df_clean) * 100).round(2)
}).sort_values('Pourcentage', ascending=False)

print(f"=== TOUTES les colonnes du dataset ({len(df_clean.columns)} colonnes) ===\n")
print(all_columns_stats.to_string(index=False))

# Colonnes complètes (sans valeurs manquantes)
colonnes_completes = all_columns_stats[all_columns_stats['Valeurs_manquantes'] == 0]
print(f"\n✅ Colonnes COMPLÈTES (sans valeurs manquantes) : {len(colonnes_completes)}")
print(colonnes_completes['Colonne'].tolist())
```

## 5. Exportation du dataset nettoyé

```python
# Sauvegarde du dataset nettoyé dans un nouveau fichier CSV
output_file = '../data/friches-standard-clean.csv'
df_clean.to_csv(output_file, sep=';', index=False, encoding='utf-8')

print(f"✅ Dataset nettoyé exporté avec succès !")
print(f"\n📁 Fichier : {output_file}")
print(f"📊 Statistiques :")
print(f"   - Lignes : {df_clean.shape[0]:,}")
print(f"   - Colonnes : {df_clean.shape[1]}")
print(f"   - Taille : {df_clean.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```
